```python
#create a sample file to load
text = '''Langchain is a framework for building Ai Apps.
RAG means Retrieval Augmented Generation.
It is a technique to retrieve relevant information from a large dataset and use it to generate new content.'''

with open("sample.txt","w") as f:
    f.write(text)

print("sample.txt is created")
```

#Step 3: load a text file with TextLoader.

```python
#Step 3: load a text file with TextLoader.

from langchain_community.document_loaders import TextLoader

# Specify the text file path
file_path = "sample.txt"

# Load the text file
loader = TextLoader(file_path, encoding="utf-8")

# Load documents
documents = loader.load()

# Print the content
for doc in documents:
    print("No of Documents:", len(documents))
    print("Document Content:")
    print(doc.page_content)

    # Print metadata
   # print(doc.metadata)
```

For Loading a pdf needs (pip install pypdf)

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(file_path)
pages = loader.load()
print(pages[0].page_content)
```
